# Experiments with Pruning Schemes

This notebook is designed to experiment with different pruning schemes on the MNIST and Fashion MNIST datasets. We will load the configurations, initialize the models, and apply various pruning techniques to evaluate their performance.

```python
import yaml
import torch
from src.utils.data_loader import load_data
from src.models.simple_cnn import SimpleCNN
from src.pruning.magnitude_pruning import MagnitudePruner
from src.pruning.fim_pruning import FIMPruner
from src.pruning.magnitude_fim_one_shot import MagnitudeFIMOneShotPruner
from src.pruning.magnitude_fim_iterative import MagnitudeFIMIterativePruner
from src.pruning.sqrt_averaged_magnitude_fim import SqrtAveragedMagnitudeFIMPruner

# Load configuration files
with open('src/config.yaml', 'r') as file:
    config = yaml.safe_load(file)

# Load datasets
train_loader, test_loader = load_data(config['dataset'])

# Initialize model
model = SimpleCNN(config['model'])

# Define pruning schemes
pruning_schemes = {
    'magnitude': MagnitudePruner(model, config['pruning']['magnitude']),
    'fim': FIMPruner(model, config['pruning']['fim']),
    'magnitude_fim_one_shot': MagnitudeFIMOneShotPruner(model, config['pruning']['magnitude_fim_one_shot']),
    'magnitude_fim_iterative': MagnitudeFIMIterativePruner(model, config['pruning']['magnitude_fim_iterative']),
    'sqrt_averaged_magnitude_fim': SqrtAveragedMagnitudeFIMPruner(model, config['pruning']['sqrt_averaged_magnitude_fim'])
}

# Experiment with each pruning scheme
results = {}
for name, pruner in pruning_schemes.items():
    print(f'Applying {name} pruning...')
    pruner.apply_pruning()
    # Here you would typically train the model and evaluate its performance
    # For demonstration, we will just store a placeholder result
    results[name] = 'Results after applying ' + name

# Display results
for name, result in results.items():
    print(f'{name}: {result}')
```
